Numpy Mini Project

```python
import numpy as np
```

```python
marks = np.array([[85,78,90,72,88],
                  [76,82,75,80,79],
                  [92,95,89,91,94],
                  [65,70,68,74,71],
                  [88,84,91,86,90],
                  [55,62,58,67,60]])
print(marks.ndim)
print(marks.shape)
```

```python
sum = np.sum(marks)
print(sum)
```

```python
total_avg = np.mean(marks)
print(total_avg)
```

```python
largest_value = np.max(marks)
print(largest_value)

lowest_value = np.min(marks)
print(lowest_value)
```

```python
position_largest = np.argmax(marks)
print(position_largest)


position_lowest = np.argmin(marks)
print(position_lowest)
```

```python
high_marks = np.where(marks >= 80)
print(high_marks)
```

```python
sorted_marks = np.sort(marks)
print(sorted_marks )
```

```python
student_average = np.mean(marks , axis =1)
print(student_average)
```

```python
subject_average = np.mean(marks , axis = 0)
print(subject_average)
```

```python
highest_subject_index = np.argmax(subject_average)
print(highest_subject_index)
```

```python
subject_lowest_index = np.argmin(subject_average)
print(subject_lowest_index)
```

```python
passed_student = np.where(student_average >=60)
print("passed students:",passed_student)
```

```python
passed_count = np.size(passed_student)
print(passed_count)
```

```python
high_performers = np.where(student_average>=80)

print(high_performers)
```

```python
high_performers_count = np.size(high_performers)
print(high_performers_count)
```
